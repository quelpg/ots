```python
from pandapower.plotting import simple_plot
import plotly.graph_objects as go
import json

# coordinates
def get_bus_coords(net):
    """Extract x,y from GeoJSON geo column."""
    coords = {}
    for idx, row in net.bus.iterrows():
        geo = row['geo']
        if geo is not None:
            g = json.loads(geo) if isinstance(geo, str) else geo
            coords[idx] = (g['coordinates'][0], g['coordinates'][1])
    return coords

coords = get_bus_coords(net)

# --- Draw line edges ---
edge_x, edge_y = [], []
for _, line in net.line.iterrows():
    x0, y0 = coords[line['from_bus']]
    x1, y1 = coords[line['to_bus']]
    edge_x += [x0, x1, None]
    edge_y += [y0, y1, None]

# Trafo edges
trafo_x, trafo_y = [], []
for _, trafo in net.trafo.iterrows():
    x0, y0 = coords[trafo['hv_bus']]
    x1, y1 = coords[trafo['lv_bus']]
    trafo_x += [x0, x1, None]
    trafo_y += [y0, y1, None]

edge_trace = go.Scatter(x=edge_x, y=edge_y, mode='lines',
    line=dict(width=1, color='gray'), hoverinfo='none', name='Lines')

trafo_trace = go.Scatter(x=trafo_x, y=trafo_y, mode='lines',
    line=dict(width=2, color='orange', dash='dash'), hoverinfo='none', name='Transformers')

# --- Draw buses colored by voltage level ---
color_map = {138.0: 'blue', 345.0: 'red', 161.0: 'green'}
for vn, color in color_map.items():
    bus_idx = net.bus[net.bus['vn_kv'] == vn].index
    bus_x = [coords[b][0] for b in bus_idx]
    bus_y = [coords[b][1] for b in bus_idx]
    # Build hover text with bus info
    hover = []
    for b in bus_idx:
        txt = f"Bus {net.bus.at[b, 'name']}<br>{vn} kV"
        # Check if there's load
        load_at_bus = net.load[net.load['bus'] == b]
        if len(load_at_bus) > 0:
            txt += f"<br>Load: {load_at_bus['p_mw'].sum():.1f} MW"
        # Check if there's gen
        gen_at_bus = net.gen[net.gen['bus'] == b]
        if len(gen_at_bus) > 0:
            txt += f"<br>Gen: {gen_at_bus['max_p_mw'].sum():.1f} MW (max)"
        if b == net.ext_grid.at[0, 'bus']:
            txt += '<br><b>SLACK</b>'
        hover.append(txt)
    locals()[f'trace_{int(vn)}'] = go.Scatter(
        x=bus_x, y=bus_y, mode='markers',
        marker=dict(size=8, color=color, line=dict(width=1, color='black')),
        text=hover, hoverinfo='text', name=f'{int(vn)} kV')

fig = go.Figure(data=[edge_trace, trafo_trace, trace_138, trace_345, trace_161])
fig.update_layout(
    title='IEEE 118-Bus System',
    showlegend=True,
    xaxis=dict(showgrid=False, zeroline=False, showticklabels=False),
    yaxis=dict(showgrid=False, zeroline=False, showticklabels=False),
    width=900, height=700,
    plot_bgcolor='white'
)
fig.show()
```

## 4. Build Unified Branch Data & Convert to Per-Unit

We combine transmission lines and transformers into a single branch list and convert all quantities to a common system base.

Base values:
- $S_{base}$ = 100 MVA (pandapower default)
- $V_{base}$ = nominal voltage of each bus (kV)

Conversions:
- Power: $P_{pu} = P_{MW} / S_{base}$
- Line ratings: $F_{max,pu} = F_{max,MVA} / S_{base}$
- Reactance: $x_{pu} = x_{\Omega} / Z_{base}$, where $Z_{base} = V_{base}^2 / S_{base}$

```python
base_mva = net.sn_mva  # system base MVA
print(f"System base: {base_mva} MVA")
print()

# --- Load demand in per-unit ---
load_p_pu = net.load["p_mw"].values / base_mva
load_q_pu = net.load["q_mvar"].values / base_mva
load_buses = net.load["bus"].values

print("Load demand (first 10):")
print(f"  Bus | P (MW)    | P (pu)    | Q (Mvar)  | Q (pu)")
for i in range(min(10, len(load_buses))):
    print(f"  {load_buses[i]:3d} | {net.load['p_mw'].iloc[i]:9.2f} | {load_p_pu[i]:9.4f} | {net.load['q_mvar'].iloc[i]:9.2f} | {load_q_pu[i]:9.4f}")

print()

# --- Generator limits in per-unit ---
gen_pmax_pu = net.gen["max_p_mw"].values / base_mva
gen_pmin_pu = net.gen["min_p_mw"].values / base_mva
gen_buses = net.gen["bus"].values

print("Generator limits (first 10):")
print(f"  Bus | Pmin (MW) | Pmin (pu) | Pmax (MW) | Pmax (pu)")
for i in range(min(10, len(gen_buses))):
    print(f"  {gen_buses[i]:3d} | {net.gen['min_p_mw'].iloc[i]:9.2f} | {gen_pmin_pu[i]:9.4f} | {net.gen['max_p_mw'].iloc[i]:9.2f} | {gen_pmax_pu[i]:9.4f}")

print()

# --- Build unified branch data ---
branches_from = []
branches_to = []
branches_x = []
branches_rate = []

# (do I keep it like this or should I vectorize it and remove the loops?)
# Transmission lines: convert x from ohm to pu
for idx, line in net.line.iterrows():
    from_bus = line["from_bus"]
    to_bus = line["to_bus"]
    vn_kv = net.bus.at[from_bus, "vn_kv"]
    z_base = (vn_kv ** 2) / base_mva
    x_pu = (line["x_ohm_per_km"] * line["length_km"]) / z_base
    rate_mva = line["max_i_ka"] * vn_kv * np.sqrt(3)  # kA * kV * sqrt(3)
    branches_from.append(from_bus)
    branches_to.append(to_bus)
    branches_x.append(x_pu)
    branches_rate.append(rate_mva)

# Transformers: convert vk_percent to pu on system base
for idx, trafo in net.trafo.iterrows():
    from_bus = trafo["hv_bus"]
    to_bus = trafo["lv_bus"]
    x_pu = (trafo["vk_percent"] / 100.0) * (base_mva / trafo["sn_mva"])
    rate_mva = trafo["sn_mva"]
    branches_from.append(from_bus)
    branches_to.append(to_bus)
    branches_x.append(x_pu)
    branches_rate.append(rate_mva)

branches_from = np.array(branches_from, dtype=int)
branches_to = np.array(branches_to, dtype=int)
branches_x = np.array(branches_x)
branches_rate = np.array(branches_rate)

num_buses = len(net.bus)
num_branches = len(branches_x)

# --- Branch thermal ratings in per-unit ---
branches_rate_pu = branches_rate / base_mva

print("Branch ratings (first 10):")
print(f"  From | To  | x (pu)    | Rate (MVA) | Rate (pu)")
for i in range(min(10, num_branches)):
    print(f"  {branches_from[i]:4d} | {branches_to[i]:3d} | {branches_x[i]:9.4f} | {branches_rate[i]:10.2f} | {branches_rate_pu[i]:9.4f}")

print(f"\nTotal branches: {num_branches}")
print(f"  Lines: {len(net.line)}")
print(f"  Transformers: {len(net.trafo)}")
print(f"\nNote: Voltages in net.bus are already in per-unit (vm_pu).")
```

## 5. DC Power Flow Matrices

We compute:
- **B** — Branch Susceptance Matrix (diagonal, $b_l = 1/x_l$)
- **A** — Incidence Matrix (branches × non-slack buses)
- **B'** — Bus Susceptance Matrix ($A^T B A$)
- **Φ (Phi)** — Sensitivity Matrix ($B \cdot A \cdot (B')^{-1}$)

```python
# Slack bus
slack_bus = net.ext_grid.at[0, "bus"] # finding slack bus
print(f"Slack bus: {slack_bus}")

all_buses = net.bus.index.values # list of all bus indexes
non_slack_buses = [b for b in all_buses if b != slack_bus] # list of all buses but the slack one
bus_to_col = {bus: idx for idx, bus in enumerate(non_slack_buses)} # maps each bus number to its column position

# Handle zero reactances
zero_x_mask = branches_x == 0
if zero_x_mask.any():
    print(f"WARNING: {zero_x_mask.sum()} branches with zero reactance (replaced with 1e-6)")
    branches_x[zero_x_mask] = 1e-6

print(f"\nMatrix dimensions:")
print(f"  B: {num_branches} x {num_branches}")
print(f"  A: {num_branches} x {num_buses - 1}")
print(f"  B': {num_buses - 1} x {num_buses - 1}")
print(f"  Phi: {num_branches} x {num_buses - 1}")
```

```python
# B: Branch Susceptance Matrix
B = np.diag(1.0 / branches_x)

# A: Incidence Matrix
A = np.zeros((num_branches, num_buses - 1))
for i in range(num_branches):
    f = branches_from[i]
    t = branches_to[i]
    if f in bus_to_col:
        A[i, bus_to_col[f]] = 1
    if t in bus_to_col:
        A[i, bus_to_col[t]] = -1

# B': Bus Susceptance Matrix
B_prime = A.T @ B @ A

# Phi: Sensitivity Matrix
B_prime_inv = np.linalg.inv(B_prime)
Phi = B @ A @ B_prime_inv

print("Pre-processing completed!")
print(f"B shape: {B.shape}")
print(f"A shape: {A.shape}")
print(f"B' shape: {B_prime.shape}")
print(f"Phi shape: {Phi.shape}")
```

## 6. Visualize Matrices

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 12))

# B matrix
axes[0, 0].spy(B, markersize=1)
axes[0, 0].set_title("B (Branch Susceptance Matrix)")

# A matrix
axes[0, 1].spy(A, markersize=1)
axes[0, 1].set_title("A (Incidence Matrix)")

# B' matrix
from matplotlib.colors import SymLogNorm
B_prime_masked = np.ma.masked_where(B_prime == 0, B_prime)
norm = SymLogNorm(linthresh=1, linscale=1, vmin=B_prime.min(), vmax=B_prime.max())
cmap_bp = plt.cm.RdBu_r.copy()
cmap_bp.set_bad(color='white')
im = axes[1, 0].imshow(B_prime_masked, aspect='auto', cmap=cmap_bp, norm=norm)
axes[1, 0].set_title("B' (Bus Susceptance Matrix)")
plt.colorbar(im, ax=axes[1, 0])

# Phi matrix - mask zeros so they appear white
Phi_masked = np.ma.masked_where(Phi == 0, Phi)
cmap_phi = plt.cm.RdBu_r.copy()
cmap_phi.set_bad(color='white')
im2 = axes[1, 1].imshow(Phi_masked, aspect='auto', cmap=cmap_phi)
axes[1, 1].set_title("Φ (Sensitivity / PTDF Matrix)")
plt.colorbar(im2, ax=axes[1, 1])

plt.tight_layout()
plt.show()
```

## 7. Save Matrices to CSV

```python
os.makedirs('matrices', exist_ok=True)

# Branch labels like '1-2', '1-3', etc.
branch_labels = [f'{branches_from[i]+1}-{branches_to[i]+1}' for i in range(num_branches)]

# Bus column labels
non_slack_buses = [b+1 for b in sorted(bus_to_col.keys())]

# B: rows and cols are branches
pd.DataFrame(B, index=branch_labels, columns=branch_labels).to_csv('data/matrices/B_branch_susceptance.csv')

# A: rows are branches, cols are non-slack buses
pd.DataFrame(A, index=branch_labels, columns=non_slack_buses).to_csv('data/matrices/A_incidence.csv')

# B': rows and cols are non-slack buses
pd.DataFrame(B_prime, index=non_slack_buses, columns=non_slack_buses).to_csv('data/matrices/B_prime_bus_susceptance.csv')

# Phi: rows are branches, cols are non-slack buses
pd.DataFrame(Phi, index=branch_labels, columns=non_slack_buses).to_csv('data/matrices/Phi_sensitivity.csv')

print('Matrices saved to data/matrices/ folder.')
```

## 8. Save Network Data to CSV

Export generation limits, load data, branch ratings and slack bus index so that
the optimisation model (`dcopf.py`) can load everything from CSV files without
depending on pandapower at run time.

```python
# --- Generator data (53 generators, no slack) ---
gen_df = pd.DataFrame({
    'bus_pp':   gen_buses,           # 0-indexed (pandapower)
    'bus_matpower': gen_buses + 1,   # 1-indexed (MATPOWER / human-readable)
    'pmin_pu':  gen_pmin_pu,
    'pmax_pu':  gen_pmax_pu,
    'pmin_mw':  gen_pmin_pu * base_mva,
    'pmax_mw':  gen_pmax_pu * base_mva,
})
gen_df.to_csv('data/gen_data.csv', index=False)

# --- Load data (99 loads) ---
load_df = pd.DataFrame({
    'bus_pp':      load_buses,
    'bus_matpower': load_buses + 1,
    'p_pu':        load_p_pu,
    'p_mw':        load_p_pu * base_mva,
})
load_df.to_csv('data/load_data.csv', index=False)

# --- Branch data (173 lines + 13 transformers = 186 branches) ---
branch_df = pd.DataFrame({
    'label':       branch_labels,
    'from_bus_pp': branches_from,          # 0-indexed
    'to_bus_pp':   branches_to,            # 0-indexed
    'from_bus_matpower': branches_from + 1,
    'to_bus_matpower':   branches_to   + 1,
    'x_pu':        branches_x,
    'rate_mva':    branches_rate,
    'rate_pu':     branches_rate_pu,
})
branch_df.to_csv('data/branch_data.csv', index=False)

# --- Slack bus index ---
slack_df = pd.DataFrame({
    'slack_bus_pp':       [slack_bus],
    'slack_bus_matpower': [slack_bus + 1],
})
slack_df.to_csv('data/slack_bus.csv', index=False)

# --- System base ---
pd.DataFrame({'base_mva': [base_mva]}).to_csv('data/base_mva.csv', index=False)

print('Network data saved to data/ folder.')
print(f'  gen_data.csv    : {len(gen_df)} generators')
print(f'  load_data.csv   : {len(load_df)} loads')
print(f'  branch_data.csv : {len(branch_df)} branches')
print(f'  slack_bus.csv   : bus {slack_bus} (pp) = bus {slack_bus+1} (MATPOWER)')
print(f'  base_mva.csv    : {base_mva} MVA')
```
